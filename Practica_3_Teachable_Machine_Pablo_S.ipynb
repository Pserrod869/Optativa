````python
# implementa la clasificación con la imagen que pertenece claramente a una categoría
import numpy as np
from PIL import Image, ImageOps

def predecir_imagen(ruta_imagen):
    # Preparar el array para el modelo
    data = np.ndarray(shape=(1, 224, 224, 3), dtype=np.float32)

    # Cargar y preprocesar la imagen
    image = Image.open(ruta_imagen).convert("RGB")
    size = (224, 224)
    image = ImageOps.fit(image, size, Image.Resampling.LANCZOS)

    # Normalizar
    image_array = np.asarray(image)
    normalized_image_array = (image_array.astype(np.float32) / 127.5) - 1
    data[0] = normalized_image_array

    # Realizar predicción
    prediction = model.predict(data)
    index = np.argmax(prediction)
    class_name = class_names[index]
    confidence_score = prediction[0][index]

    print(f"Imagen: {ruta_imagen}")
    print("Class:", class_name[2:].strip())
    print("Confidence Score:", confidence_score)
    print("Distribución de probabilidad completa:")
    for idx, name in enumerate(class_names):
        print(f"  {name.strip()}: {prediction[0][idx]:.4f}")

# Reemplaza 'clara.png' con el nombre de tu archivo de imagen clara
try:
    predecir_imagen('clara.png')
except Exception as e:
    print(f"Error: Asegúrate de haber subido la imagen 'clara.png'. Detalle: {e}")
````

````python
# implementa la clasificación con una imagen ambigua, que puede ser de cualquier categoría
# Reemplaza 'ambigua.png' con el nombre de tu archivo de imagen ambigua
try:
    predecir_imagen('ambigua.png')
except Exception as e:
    print(f"Error: Asegúrate de haber subido la imagen 'ambigua.png'. Detalle: {e}")
````

````python
# implementa la clasificación con una imagen sin ninguna categoría del modelo
# Reemplaza 'otra.png' con el nombre de tu archivo de imagen que no pertenece a ninguna categoría
try:
    predecir_imagen('otra.png')
except Exception as e:
    print(f"Error: Asegúrate de haber subido la imagen 'otra.png'. Detalle: {e}")
````

### ¿Cómo uso mi modelo exportado?

Para usar el modelo que descargué de Teachable Machine, sigo estos pasos simples en mi código:

1. **Cargar el modelo y las etiquetas**: Uso `keras.models.load_model` para cargar el archivo del modelo (`keras_model.h5`) y abro el archivo `labels.txt` para saber qué nombres corresponden a cada predicción.
2. **Preparar la imagen**: El modelo solo entiende imágenes cuadradas de $224 \times 224$ píxeles. Por eso, redimensiono mi foto a ese tamaño y la convierto en números (una matriz de NumPy).
3. **Normalizar**: Divido los valores de la imagen para que estén en un rango de -1 a 1, que es como el modelo aprendió a procesar los datos durante su entrenamiento.
4. **Predicción**: Le paso la imagen procesada al modelo con `model.predict()`. Esto me devuelve una lista de probabilidades para cada clase. Me quedo con la clase que tenga el porcentaje más alto (usando `np.argmax`).

---

### ¿Cómo se comporta mi modelo?

* **Con imágenes claras**: El modelo funciona muy bien. Si le paso una foto nítida y bien centrada de un objeto con el que entrenó, me da la categoría correcta con una confianza casi del 100% (cercana a `1.0`).
* **Con imágenes dudosas o ambiguas**: Si la foto tiene mala luz, está borrosa o muestra una mezcla de objetos, el modelo duda. En este caso, reparte los porcentajes entre varias categorías y la confianza de la predicción ganadora baja bastante (por ejemplo, a `0.5` o `0.6`).
* **Con cosas que no conoce**: El modelo no tiene una opción para decir "no sé qué es esto". Como la suma de todos los porcentajes siempre tiene que dar 100%, si le enseño un objeto totalmente nuevo que nunca vio en el entrenamiento, se verá obligado a clasificarlo en la categoría que le parezca más similar visualmente (por colores o formas), a veces con una confianza engañosamente alta.
